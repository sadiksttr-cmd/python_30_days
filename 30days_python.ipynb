```python
base = input ('Please enter the base of the triangle:')
height = input ('Please enter the height of the triangle')

try:
    base_float = float(base)
    height_float = float(height)
    print('Area of the trianlge is:', base_float*height_float*0.5)

except ValueError:
    print('inputs musht be numbers')
```

```python
a = input ('Side a: ')
b = input('Side b: ')
c = input('Side c')

try:
    side_a = float (a)
    side_b = float (b)
    side_c = float (c)
    print('Sum of the three input values', sum([side_a, side_b,side_c]))
except ValueError:
    print('inouts should be numbers')
```
